**MISSING DATA**

```python
#FINDING MISSING DATA

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df  #NaN = not a number -> Null values
```

```python
df.isna()  #if it is a null value == True, else false
```

.sum(): When applied to the boolean DataFrame returned by df.isna(), the sum() method counts the True values in each column. Since True is treated as 1 and False as 0 in numerical operations, summing the boolean values effectively counts the number of missing values in each column.

```python
df.isna().sum()
```

```python
df
```

**Merging 2 dataframes**

Use When:

You have related data in different tables.

You want to combine them using a key (like id or name).

```python
 employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on='employee_id',how ='inner')
```

```python
pd.merge(employees,salaries,on='employee_id',how ='outer')
```

```python
pd.merge(employees,salaries,on='employee_id',how ='left')  #based on employees
```

```python
pd.merge(employees,salaries,on='employee_id',how ='right')  #based on salaries
```

**Pivot Tables**

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values = "Sales",index = 'Region',columns="Product",aggfunc = 'median')
```

```python
pivot2 = pd.pivot_table(df, values=['Sales', 'Units'], index='Region', columns='Product')
pivot2
```

**Cross Tabs**

```python
pd.crosstab(df['Region'],df['Product']) #count
```
