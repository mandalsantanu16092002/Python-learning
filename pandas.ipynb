**MISSING DATA**

```python
#FINDING MISSING DATA

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df  #NaN = not a number -> Null values
```

```python
df.isna()  #if it is a null value == True, else false
```

.sum(): When applied to the boolean DataFrame returned by df.isna(), the sum() method counts the True values in each column. Since True is treated as 1 and False as 0 in numerical operations, summing the boolean values effectively counts the number of missing values in each column.

```python
df.isna().sum()
```

```python
df
```

**Removing Missing Data**

```python
df.dropna()
```

```python
df.dropna(thresh=3) #rows a non null value 3 ta thakbe
```

**Filling the missing data**

```python
df.fillna(0)  #sob NaN = 0 hoye jabe
```

```python
values = {'A':0,'B':100,"C":300,'D':400}
df.fillna(value=values)  #column wise sob A column ar NaN te 0, sob B column ar NaN te 100
```

```python
df.fillna(df.mean())  # Column A te 1+2+4+5/4 = 3 in NaN, Column C te 1+2+3/3 = 2 in NaN, Column D te 1+5/2 = 3 in NaN
```
